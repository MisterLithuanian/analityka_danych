## Exercise 4
"I am interested in the house with 3 bedrooms, 2 bathrooms and 2 floors"  
a) Compute the basic statistics of apartments fulfilling above constraints.  
b) Distinct accepted apartments on any of previously created plots.  
c) Find such apartments with best price to footage ratio.  

```python
#a)
#Your code:

houses = df.loc[
    (df["bedrooms"] == 3) &
    (df["bathrooms"] == 2) &
    (df["floors"] == 2)
]

print("Number of matching houses:", len(houses))
houses.describe()
#Hint: pandas.DataFrame.loc
```

```python
#b)

plt.figure(figsize=(8, 5))


plt.scatter(df["sqft_living"], df["price"], label="All houses")

plt.scatter(houses["sqft_living"], houses["price"], color="red", label="3 bedrooms, 2 bathrooms, 2 floors")

plt.xlabel("Living area (sqft)")
plt.ylabel("Price")
plt.title("Selected Houses vs All Houses")
plt.legend()
plt.show()
```

```python
#c)
houses = houses.copy()
houses["price_per_sqft"] = houses["price"] / houses["sqft_living"]

best_houses = houses.sort_values("price_per_sqft").head(10)

print(best_houses[["id", "price", "sqft_living", "price_per_sqft", "bedrooms", "bathrooms", "floors", "grade", "zipcode"]])
```
